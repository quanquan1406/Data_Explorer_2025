# Tiền xử lý dữ liệu

```python
import pandas as pd
import numpy as np
```

# File 2.1.5 (live & his) industries_performance_processed.csv 

```python
# Đọc dữ liệu từ file CSV
file_path_215 = "./DATA EXPLORER CONTEST/Data - 12ndMar/2.1.5 (live & his) industries_performance_processed.csv"
df_215 = pd.read_csv(file_path_215)

# Xem thông tin tổng quan
print("Kích thước dữ liệu:", df_215.shape)
print("\nCác cột dữ liệu:\n", df_215.dtypes)
print("\nDữ liệu bị thiếu:\n", df_215.isnull().sum())
print("\n5 dòng đầu tiên của dữ liệu:")
print(df_215.head())
```

```python
# Hàm chuyển từ % string → float
def convert_percent(val):
    if isinstance(val, str):
        val = val.replace(',', '')  # Xóa dấu phẩy hàng nghìn
        val = val.replace('%', '')  # Xóa ký hiệu %
        try:
            return float(val)
        except:
            return np.nan
    return val

# Áp dụng chuyển đổi cho tất cả cột ngoại trừ 'Ngành'
cols_to_convert = df_215.columns[df_215.columns != 'Ngành']
df_215[cols_to_convert] = df_215[cols_to_convert].applymap(convert_percent)

# Xử lý missing values
# Điền bằng giá trị trung bình
df_215.fillna(df_215.mean(numeric_only=True), inplace=True)


# Kiểm tra lại kết quả
print("Kích thước dữ liệu:", df_215.shape)
print("\nCác cột dữ liệu:\n", df_215.dtypes)
print("\nDữ liệu bị thiếu:\n", df_215.isnull().sum())
print("\n5 dòng đầu tiên của dữ liệu:")
print(df_215.head())
```

# File 2.2.1 (live & his) detail_industry_performance_processed.csv 

```python
# Đọc dữ liệu từ file CSV
file_path_221 = "./DATA EXPLORER CONTEST/Data - 12ndMar/2.2.1 (live & his) detail_industry_performance_processed.csv"
df_221 = pd.read_csv(file_path_221)

# Xem thông tin tổng quan
print("Kích thước dữ liệu:", df_221.shape)
print("\nCác cột dữ liệu:\n", df_221.dtypes)
print("\nDữ liệu bị thiếu:\n", df_221.isnull().sum())
print("\n5 dòng đầu tiên của dữ liệu:")
print(df_221.head())
```

```python
# Chuyển đổi các cột tỷ lệ phần trăm từ M10_2010 đến W9_2025
percentage_columns = df_221.columns[2:]  # Các cột từ M10_2010 đến W9_2025

# Hàm chuyển đổi tỷ lệ phần trăm sang số thực
def convert_to_float(x):
    if isinstance(x, str):
        return float(x.replace('%', '')) / 100
    return x

# Áp dụng hàm convert_to_float cho tất cả các cột tỷ lệ phần trăm
df_221[percentage_columns] = df_221[percentage_columns].applymap(convert_to_float)


# Bước 1: Chuyển đổi dữ liệu từ dạng rộng thành dài (cột thành hàng)
df_melted = df_221.melt(id_vars=['Industry', 'Type of Value'], 
                        value_vars=percentage_columns, 
                        var_name='Period', 
                        value_name='Performance')

# Bước 2: Chuyển đổi các giá trị phần trăm thành kiểu số thực (nếu chưa thực hiện ở bước trên)
df_melted['Performance'] = df_melted['Performance'].apply(convert_to_float)

# Kiểm tra kết quả
print("\nDữ liệu sau khi chuyển đổi từ cột thành hàng:")
print(df_melted.head(10))
```

# File 3.4 (live & his) news_internal_transactions (FPT_CMG)_processed.csv 

```python
# Đọc dữ liệu từ file CSV
file_path_34 = "./DATA EXPLORER CONTEST/News - FPT & CMG/3.4 (live & his) news_internal_transactions (FPT_CMG)_processed.csv"
df_34 = pd.read_csv(file_path_34)

# Xem thông tin tổng quan
print("Kích thước dữ liệu:", df_34.shape)
print("\nCác cột dữ liệu:\n", df_34.dtypes)
print("\nDữ liệu bị thiếu:\n", df_34.isnull().sum())
print("\n5 dòng đầu tiên của dữ liệu:")
print(df_34.head())
```

```python
# Bước 1: Xử lý giá trị thiếu (có thể thay thế hoặc loại bỏ)
# Thay thế các giá trị thiếu trong các cột cần thiết bằng 'Unknown' hoặc 'NaN'
df_34['Executor Name'].fillna('Unknown', inplace=True)
df_34['Executor Position'].fillna('Unknown', inplace=True)
df_34['Related Person Name'].fillna('Unknown', inplace=True)
df_34['Related Person Position'].fillna('Unknown', inplace=True)
df_34['Relation'].fillna('Unknown', inplace=True)

# Bước 2: Chuyển đổi các cột ngày tháng sang kiểu datetime
df_34['Registered From Date'] = pd.to_datetime(df_34['Registered From Date'], errors='coerce', format='%d/%m/%Y')
df_34['Registered To Date'] = pd.to_datetime(df_34['Registered To Date'], errors='coerce', format='%d/%m/%Y')
df_34['Executed From Date'] = pd.to_datetime(df_34['Executed From Date'], errors='coerce', format='%d/%m/%Y')
df_34['Executed To Date'] = pd.to_datetime(df_34['Executed To Date'], errors='coerce', format='%d/%m/%Y')

# Bước 3: Chuyển đổi các cột giao dịch từ chuỗi sang số (loại bỏ dấu phẩy và chuyển thành số)
def convert_to_numeric(column):
    return pd.to_numeric(column.replace({',': ''}, regex=True), errors='coerce')

df_34['Before Transaction Volume'] = convert_to_numeric(df_34['Before Transaction Volume'])
df_34['Registered Transaction Volume'] = convert_to_numeric(df_34['Registered Transaction Volume'])
df_34['Executed Transaction Volume'] = convert_to_numeric(df_34['Executed Transaction Volume'])
df_34['After Transaction Volume'] = convert_to_numeric(df_34['After Transaction Volume'])

# Bước 4: Xử lý các cột tỷ lệ phần trăm (có thể thay thế NaN bằng giá trị trung bình hoặc 0)
df_34['Before Transaction Percentage'].fillna(df_34['Before Transaction Percentage'].mean(), inplace=True)
df_34['After Transaction Percentage'].fillna(df_34['After Transaction Percentage'].mean(), inplace=True)

# Bước 5: Kiểm tra lại dữ liệu sau khi xử lý
print("Kích thước dữ liệu sau khi xử lý:", df_34.shape)
print("\nDữ liệu bị thiếu sau khi xử lý:\n", df_34.isnull().sum())
print("\n5 dòng đầu tiên của dữ liệu sau khi xử lý:")
print(df_34.head())
```

```python
# Bước 1: Làm sạch số liệu (bỏ dấu phẩy, chuyển về số)
value_cols = df_61.columns.difference(['Indicator', 'StockID'])
df_61[value_cols] = df_61[value_cols].replace({',': ''}, regex=True)
df_61[value_cols] = df_61[value_cols].apply(pd.to_numeric, errors='coerce')

# Bước 2: Pivot - mỗi dòng là 1 timepoint, mỗi cột là 1 Indicator cho từng công ty
df_61_pivoted = df_61.pivot_table(
    index='StockID',
    columns='Indicator',
    values=value_cols
)

# Đưa multi-index về dạng bình thường
df_61_pivoted = df_pivoted.T  # Transpose để đưa thời gian lên hàng
df_61_pivoted.index.name = 'TimePeriod'
df_61_pivoted.reset_index(inplace=True)

# Xem dữ liệu sau xử lý
print("Kích thước sau pivot:", df_61_pivoted.shape)
print("5 dòng đầu:")
print(df_61_pivoted.head())
```
