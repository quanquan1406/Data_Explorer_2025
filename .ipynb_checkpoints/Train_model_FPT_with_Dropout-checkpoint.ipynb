```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler

from keras.models import Sequential
from keras.layers import Dense, Dropout, LSTM
```

```python
# Đọc dữ liệu từ file CSV
file_path = "./DATA EXPLORER CONTEST/Data - 12ndMar/4.2.3 (TARGET) (live & his) FPT_detail_transactions_processed.csv"
df = pd.read_csv(file_path)

# Xem thông tin tổng quan
print("Kích thước dữ liệu:", df.shape)
print("\nCác cột dữ liệu:\n", df.dtypes)
print("\nDữ liệu bị thiếu:\n", df.isnull().sum())
print("\n5 dòng đầu tiên của dữ liệu:")
print(df.head())
```

```python
# Chuyển đổi cột 'Date' sang datetime
df["Date"] = pd.to_datetime(df["Date"], format="%d/%m/%Y")

# Hàm chuyển đổi số có dấu ','
def convert_numeric(column):
    return df[column].astype(str).str.replace(",", "").astype(float)

# Chuyển đổi các cột số sang dạng float
num_cols = ["Total Volume", "Total Value", "Market Cap", "Closing Price", 
            "Price Change", "Matched Volume", "Matched Value"]
for col in num_cols:
    df[col] = convert_numeric(col)

# Kiểm tra lại kiểu dữ liệu
print(df.dtypes)
print(df.head())
```

# Long Short Term Memory (LSTM)

```python
# Sắp xếp dữ liệu theo thời gian
df = df.sort_values(by="Date")

# Tạo DataFrame và đặt Date làm index
Cls_price_FPT = df[['Date', 'Closing Price']].set_index('Date')
Cls_price_FPT
```
